```python
from pptx import Presentation
from pptx.util import Inches, Pt
from pptx.enum.text import PP_ALIGN
from pptx.enum.text import MSO_AUTO_SIZE
import os
import comtypes.client

# Path to the folder containing charts
img_folder = "makeup_visualizations_png"

# Create a presentation
prs = Presentation()
prs.slide_width = Inches(16)
prs.slide_height = Inches(9)

# List of slides
slides_info = [
    {"filename": "1_Number_of_unique_values.png", "title": "Number of Unique Products, Brands, Categories, and Types",
     "description": "• Chart type: barh (horizontal bar chart).\n\n• Displayed: Number of unique values for each data category."},

    {"filename": "2_Number_brands_products_category_product_type.png", "title": "Number of Brands and Products by Categories",
     "description": "• Chart type: barh (horizontal bars).\n\n• Displayed:\n\n  – Brand — Number of unique brands (on the left).\n\n  – Name — Number of unique products (on the right) for each combination of category and product type."},

    {"filename": "3_4_5_6_Price.png", "title": "Price Analysis",
     "description": "• Chart types: bar, pie, boxplot, scatter.\n\n• Displayed:\n\n  – Bar chart: Number of items by price categories.\n\n  – Pie chart: Percentage share of each price category.\n\n  – Boxplot: Minimum, maximum, median price, quartiles.\n\n  – Scatter plot: Dependency between price and rating."},

    {"filename": "7_8_9_Rating.png", "title": "Product Ratings",
     "description": "• Chart types: kdeplot (density estimation), pie, bar.\n\n• Displayed:\n\n  – Distribution of rating values.\n\n  – Share of items with and without ratings.\n\n  – Count of types, categories, and brands with ratings."},

    {"filename": "10_Product_Type.png", "title": "Product Type Analysis",
     "description": "• Chart types: bar (bars), line (lines).\n\n• Displayed:\n\n  – Number of products for each type.\n\n  – Average rating and price by product types."},

    {"filename": "11_Category.png", "title": "Product Categories",
     "description": "• Chart types: bar (bars), line (lines).\n\n• Displayed:\n\n  – Number of products.\n\n  – Average rating and price by categories."},

    {"filename": "12_Product_Types_and_Categories_Sankey.png", "title": "Connections Between Product Types and Categories",
     "description": "• Chart type: Sankey Diagram.\n\n• Displayed:\n\n  – Relationships between product types and categories.\n\n  – Distribution of products between types and categories represented by line widths."},

    {"filename": "13_Brands_Categories_Product_Type_Ratings_Treemap.png", "title": "Brands by Categories and Product Types",
     "description": "• Chart type: Treemap.\n\n• Displayed:\n\n  – Hierarchical structure of brands in categories.\n\n  – Rating indicated by a color scale."},

    {"filename": "14_All_brands.png", "title": "Overview of All Brands",
     "description": "• Chart types: bar, line.\n\n• Displayed:\n\n  – Number of products for each brand.\n\n  – Average rating and price."},

    {"filename": "15_16_Top_Brands_in_Product.png", "title": "Top 20 Brands",
     "description": "• Chart types: donut chart, pie chart.\n\n• Displayed:\n\n  – Share of products offered by the top 20 brands.\n\n  – Count of categories, product types, and their coverage."},

    {"filename": "17_Top_Brands_analysis.png", "title": "Analysis of Top 20 Brands",
     "description": "• Chart types: bar, line.\n\n• Displayed:\n\n  – Number of products, ratings, and average price per brand."},

    {"filename": "18_Top_Brands_Price_statistics.png", "title": "Top Brands Prices",
     "description": "• Chart type: boxplot.\n\n• Displayed:\n\n  – Price distribution for each brand, including minimum, maximum, median."},

    {"filename": "19_Top_Brands_Price_Categor.png", "title": "Price Distribution Among Products",
     "description": "• Chart type: stacked bar.\n\n• Displayed:\n\n  – Representation of brands across different price categories."},

    {"filename": "20_Top_Brands_Across_Category_Product_Heatmap.png",
     "title": "Product Distribution Across Categories and Types Among Top 20 Brands",
     "description": "• Chart type: heatmap.\n\n• Displayed:\n\n  – Count distribution of category-product types among top 20 brands.\n\n  – The chart highlights which brands actively operate across different category-product types and their relationships."}
]

# Add slides
for info in slides_info:
    slide = prs.slides.add_slide(prs.slide_layouts[6])

    # Title
    title_box = slide.shapes.add_textbox(Inches(0.5), Inches(0.2), Inches(15), Inches(1))
    title_frame = title_box.text_frame
    title_frame.text = info["title"]
    title_frame.paragraphs[0].font.size = Pt(28)
    title_frame.paragraphs[0].alignment = PP_ALIGN.LEFT

    # Add image
    img_path = os.path.join(img_folder, info["filename"])
    if os.path.exists(img_path):
        slide.shapes.add_picture(img_path, Inches(0.5), Inches(1.2), width=Inches(10))
    else:
        print(f"File not found: {img_path}")

    # Description
    desc_box = slide.shapes.add_textbox(Inches(11), Inches(1.2), Inches(4.5), Inches(6))
    desc_frame = desc_box.text_frame
    desc_frame.clear()  # Clear the default paragraph
    desc_frame.word_wrap = True
    desc_frame.auto_size = MSO_AUTO_SIZE.TEXT_TO_FIT_SHAPE

    for line in info["description"].split("\n\n"):
        p = desc_frame.add_paragraph()
        p.text = line.strip()
        p.font.size = Pt(18)

# Save the presentation
pptx_filename = "Makeup_analysis_presentation.pptx"
prs.save(pptx_filename)
print(f"Presentation saved as {pptx_filename}")

# Convert to PDF
def pptx_to_pdf(input_file, output_file):
    powerpoint = comtypes.client.CreateObject("PowerPoint.Application")
    powerpoint.Visible = 1
    deck = powerpoint.Presentations.Open(os.path.abspath(input_file))
    deck.SaveAs(os.path.abspath(output_file), 32)
    deck.Close()
    powerpoint.Quit()

pdf_filename = "Makeup_analysis_presentation.pdf"
pptx_to_pdf(pptx_filename, pdf_filename)
print(f"Presentation saved as {pdf_filename}")
```
