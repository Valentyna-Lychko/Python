```python
import requests  # Library for making HTTP requests.
import json      # Library for handling JSON data.
import pandas as pd  # Powerful library for data analysis and manipulation.

# Load data from a previously saved JSON file.
file_path = "makeup_all_products.json"  # Specify the path to the data file.
df = pd.read_json(file_path)  # Read the JSON file into a Pandas DataFrame for analysis.

# Display summary information about the DataFrame.
print("Summary of the dataset:")
print(df.info())  # Provides row and column counts, data types, and memory usage.
```

```python
# Display the list of columns in the DataFrame
columns = df.columns.tolist()  # Retrieve the list of column names
print("List of columns in the DataFrame:")
print(columns)
```

```python
# Check the number of NaN values, empty strings, and zero values in the dataset.

# Import necessary libraries.
# import pandas as pd  # Library for data manipulation and analysis.
# import json          # Library for handling JSON files.

# File path to the data source.
# file_path = "makeup_all_products.json"

# Load the data from the specified file.
with open(file_path, "r", encoding="utf-8") as f:
    data = json.load(f)

# Convert the JSON data into a Pandas DataFrame for analysis.
df = pd.DataFrame(data)

# Prepare the summary results.
result = []

# Iterate through each column in the DataFrame to calculate statistics.
for col in df.columns:
    total_nan = df[col].isna().sum()  # Count the number of NaN values.
    total_empty_str = (df[col] == "").sum()  # Count the number of empty string values.
    total_zeros = df[col][pd.to_numeric(df[col], errors="coerce") == 0].count() if pd.api.types.is_numeric_dtype(df[col]) or df[col].dtype == object else 0  # Count the number of zero values, if applicable.

    # Append the results for the current column.
    result.append({
        "column": col,
        "NaN_count": total_nan,
        "empty_string_count": total_empty_str,
        "zero_count": total_zeros
    })

# Convert the result into a structured summary table.
summary_df = pd.DataFrame(result)

# Display the summary table, sorted by counts of NaN, empty strings, and zeros.
print(summary_df.sort_values(by=["NaN_count", "empty_string_count", "zero_count"], ascending=False))
```

```python
# Statistical overview of all numeric columns.

# Convert columns to numeric format if necessary.
# Ensures columns containing textual data are coerced into numeric format.
df['price'] = pd.to_numeric(df['price'], errors='coerce')
df['rating'] = pd.to_numeric(df['rating'], errors='coerce')

# Generate statistical information for all numeric columns.
# Calculates key metrics like count, mean, standard deviation, min, max, and quartiles.
numeric_stats = df.describe()

# Display the results in the console.
# Outputs statistical insights for numeric data in the DataFrame.
print("\nStatistical information for all numeric columns:")
print(numeric_stats)
```

```python
# Calculate the number of unique values for columns 'id', 'name', 'brand', 'category', and 'product_type'.
# Use the `nunique` function to count unique values for each specified column.
unique_counts = {
    'id': df['id'].nunique(),
    'name': df['name'].nunique(),
    'brand': df['brand'].nunique(),
    'category': df['category'].nunique(),
    'product_type': df['product_type'].nunique()
}

# Output the results.
# Display the count of unique values for each column in a structured format.
print("Number of unique values in each column:")
for column, count in unique_counts.items():
    print(f"{column}: {count}")
```

```python
# Identify rows with duplicate 'name' values and display relevant columns.

# Count occurrences of each 'name'.
# Determine how many times each 'name' appears in the DataFrame.
name_counts = df['name'].value_counts()

# Filter rows with duplicate 'name' values.
# Keep only the rows where 'name' appears more than once.
repeated_names = name_counts[name_counts > 1].index  # List of duplicated 'name' values.
filtered_df = df[df['name'].isin(repeated_names)]  # Filter DataFrame to include rows with repeated 'name' values.

# Sort the filtered rows by 'name'.
# Arrange the DataFrame in ascending order based on the 'name' column.
sorted_df = filtered_df.sort_values(by='name')

# Select relevant columns for the result.
result = sorted_df[['id', 'name', 'brand', 'category', 'product_colors', 'description', 'price', 'rating']]

# Output the results.
print("Rows sorted by 'name' where 'name' appears more than once (id, name, brand, category, product_colors, description, price, rating):")
print(result)

# Count the total number of repeated 'name' values and the number of unique duplicated 'name'.
print(f"\nTotal number of repeated 'name' values: {len(repeated_names)}")
print(f"Number of unique duplicated 'name': {len(result['name'].unique())}")
```

```python
import pandas as pd

# Import data from a JSON file.
file_path = "makeup_all_products.json"  # Specify the path to your JSON file.
df = pd.read_json(file_path)  # Load the data into a Pandas DataFrame.

# Filter data where the price is greater than 0.
filtered_prices = df[df['price'] > 0]['price']

# Calculate basic statistical metrics.
average_price = filtered_prices.mean()  # Calculate the mean price.
min_price = filtered_prices.min()       # Identify the minimum price.
max_price = filtered_prices.max()       # Identify the maximum price.

# Print basic statistics.
print("Basic price statistics (excluding 0):")
print(f"Average price: {average_price}")
print(f"Minimum price: {min_price}")
print(f"Maximum price: {max_price}")
print()

# Calculate percentiles for price distribution.
percentiles = {
    "10%": filtered_prices.quantile(0.10),
    "25%": filtered_prices.quantile(0.25),
    "50% (Median)": filtered_prices.quantile(0.50),
    "75%": filtered_prices.quantile(0.75),
    "90%": filtered_prices.quantile(0.90),
    "95%": filtered_prices.quantile(0.95),
    "99%": filtered_prices.quantile(0.99)
}

# Print percentile statistics.
print("Price percentiles:")
for label, value in percentiles.items():
    print(f"{label}: {value}")
print()

# Categorize prices into ranges based on 25% and 75% percentiles.
count_less_25 = filtered_prices[filtered_prices < percentiles["25%"]].count()
count_25_to_75 = filtered_prices[(filtered_prices >= percentiles["25%"]) & (filtered_prices <= percentiles["75%"])].count()
count_more_75 = filtered_prices[filtered_prices > percentiles["75%"]].count()

df_percentile_25_75 = pd.DataFrame({
    "Range": ["< 25%", "25%-75%", "> 75%"],
    "Count": [count_less_25, count_25_to_75, count_more_75],
    "Percentile Values": [
        f"< {percentiles['25%']}",
        f"{percentiles['25%']} to {percentiles['75%']}",
        f"> {percentiles['75%']}"
    ]
})

print("Distribution by ranges (25%-75%):")
print(df_percentile_25_75)
print()

# Categorize prices into ranges based on 10% and 90% percentiles.
count_less_10 = filtered_prices[filtered_prices < percentiles["10%"]].count()
count_10_to_90 = filtered_prices[(filtered_prices >= percentiles["10%"]) & (filtered_prices <= percentiles["90%"])].count()
count_more_90 = filtered_prices[filtered_prices > percentiles["90%"]].count()

df_percentile_10_90 = pd.DataFrame({
    "Range": ["< 10%", "10%-90%", "> 90%"],
    "Count": [count_less_10, count_10_to_90, count_more_90],
    "Percentile Values": [
        f"< {percentiles['10%']}",
        f"{percentiles['10%']} to {percentiles['90%']}",
        f"> {percentiles['90%']}"
    ]
})

print("Distribution by ranges (10%-90%):")
print(df_percentile_10_90)
print()

# Count the number of items above specific percentiles (90%, 95%, and 99%).
extreme_counts = {
    "90%": filtered_prices[filtered_prices > percentiles["90%"]].count(),
    "95%": filtered_prices[filtered_prices > percentiles["95%"]].count(),
    "99%": filtered_prices[filtered_prices > percentiles["99%"]].count()
}

print("Number of items where the price exceeds specified percentiles:")
for label, count in extreme_counts.items():
    print(f"> {label} percentile: {count}")
```
