```python
import requests
import json

# URL to the API (loading all data without filtering)
url = "https://makeup-api.herokuapp.com/api/v1/products.json"

try:
    # Loading data from the API
    response = requests.get(url)
    response.raise_for_status()  # Check for errors

    # Converting data to JSON format
    data = response.json()

    # Saving data to a file with a new name
    with open("makeup_all_products.json", "w", encoding="utf-8") as file:
        json.dump(data, file, ensure_ascii=False, indent=4)
    
    print("Data successfully saved to 'makeup_all_products.json'")
except requests.exceptions.RequestException as e:
    print(f"Error occurred while fetching data: {e}")
```
